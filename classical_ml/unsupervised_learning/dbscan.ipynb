```python
from sklearn.cluster import DBSCAN
from sklearn.datasets import make_moons
import numpy as np 
import matplotlib.pyplot as plt
```

```python
x , y = make_moons(n_samples= 1000 , noise=0.05 , random_state=42)
```

```python
print(make_moons.__doc__)
```

```python
print(x.shape , y.shape)
```

```python
plt.scatter(x[: , 0] , x[: ,1] , c=y , cmap="RdBu")
plt.show()
```

```python
#dbscan on the makemoons dataset

dbscan = DBSCAN(eps=0.05 , min_samples=5)
dbscan.fit(x)
```

```python
dbscan.labels_
```

```python
dbscan.core_sample_indices_
```

```python
dbscan.components_
```

```python
def plot_decision(x , eps , ax):
    dbscan = DBSCAN(eps=eps , min_samples=5)
    clusters = dbscan.fit_predict(x)
    ax.scatter(x[: ,0] , x[: , 1] , c=clusters , s=10)
    anomolies = x[clusters == -1]
    ax.plot(anomolies[: ,0] , anomolies[: , 1] , "rx")
    ax.grid()
    ax.set_title(f"epsilon : {eps} , min_samples : 5")
    
    
fig , ax = plt.subplots(ncols=2 , nrows=1 , figsize=(10,5))
plot_decision(x , eps=0.05 , ax=ax[0])
plot_decision(x , eps=0.20 , ax=ax[1])
plt.show()
    
    
```
